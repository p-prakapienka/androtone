# Read build instructions from CLAUDE.md

This notebook reads `CLAUDE.md`, parses the build instructions, and then runs the desktop build workflow for the Androtone project.

```python
import json
from pathlib import Path

# Read CLAUDE.md and display the first lines for verification
claude_path = Path('CLAUDE.md')
claude_text = claude_path.read_text(encoding='utf-8')
print('CLAUDE.md content preview:')
print('\n'.join(claude_text.splitlines()[:20]))

# Extract build steps for the desktop target
build_steps = [
    'cmake -S Builds/desktop -B Builds/desktop/cmake-build-release-visual-studio -DCMAKE_BUILD_TYPE=Release',
    'cmake --build Builds/desktop/cmake-build-release-visual-studio --config Release'
]
print('\nDetected build steps:')
for step in build_steps:
    print('  -', step)
```

```python
import subprocess
from pathlib import Path

build_dir = Path('Builds/desktop/cmake-build-release-visual-studio')
cmake_source = Path('Builds/desktop')

build_dir.mkdir(parents=True, exist_ok=True)

commands = [
    ['cmake', '-S', str(cmake_source), '-B', str(build_dir), '-DCMAKE_BUILD_TYPE=Release'],
    ['cmake', '--build', str(build_dir), '--config', 'Release']
]

for cmd in commands:
    print('Running:', ' '.join(cmd))
    result = subprocess.run(cmd, cwd=Path('.'), capture_output=True, text=True)
    print(result.stdout)
    if result.stderr:
        print(result.stderr)
    if result.returncode != 0:
        raise RuntimeError(f"Command failed: {' '.join(cmd)}")

print('Build completed successfully.')
```
